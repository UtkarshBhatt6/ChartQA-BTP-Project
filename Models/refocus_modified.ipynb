```python
from PIL import Image
import requests
from transformers import AutoModelForCausalLM
from transformers import AutoProcessor

# model_id = "microsoft/Phi-3.5-vision-instruct"
model_id = "Fiaa/ReFocus"
subfolder = "lr1e-06_ep2_bb1_edit0_aug0"
# Note: set _attn_implementation='eager' if you don't have flash_attn installed
model = AutoModelForCausalLM.from_pretrained(
  model_id,
   subfolder=subfolder,
  device_map="cuda",
  trust_remote_code=True,
  torch_dtype="auto",
  # _attn_implementation='flash_attention_2'
)

# for best performance, use num_crops=4 for multi-frame, num_crops=16 for single-frame.
processor = AutoProcessor.from_pretrained(model_id,  subfolder=subfolder,
  trust_remote_code=True,
  num_crops=4
)
```

```python
# 1. Create images and placeholder
images = []
placeholder = ""

for i in range(1, 3):
    url = f"https://image.slidesharecdn.com/azureintroduction-191206101932/75/Introduction-to-Microsoft-Azure-Cloud-{i}-2048.jpg"
    images.append(Image.open(requests.get(url, stream=True).raw))
    placeholder += f"<|image_{i}|>\n"

# 2. Create message
messages = [
    {"role": "user", "content": placeholder + "Summarize the deck of slides."},
]

# 3. Create prompt (text only)
prompt = processor.tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

# 4. Process (prompt + images) to inputs
inputs = processor(prompt, images=images, return_tensors="pt").to(model.device)

# 5. Generate
generation_args = {
    "max_new_tokens": 1000,
    "temperature": 0.0,
    "do_sample": False,
}
generate_ids = model.generate(
    **inputs,
    eos_token_id=processor.tokenizer.eos_token_id,
    **generation_args
)

# 6. Remove prompt part
generate_ids = generate_ids[:, inputs['input_ids'].shape[1]:]

# 7. Decode
response = processor.batch_decode(
    generate_ids,
    skip_special_tokens=True,
    clean_up_tokenization_spaces=False
)[0]

print(response)
```

```python
import os
from google import genai

model_name = "gemini-1.5-flash"
text_prompt = "What was the female population of Tunisia in 2019?"
client = genai.Client(api_key="AIzaSyCpnvgv1xsRMm5Mi34Jevpbm6rp8YQruUk")
image_paths = [
    "raw_image.png",
    "zoomed_region.jpg",
]

uploaded_files = []
print(f"Uploading {len(image_paths)} files...")
for path in image_paths:
    print(f"  Uploading: {path}")
    # Check if file exists before uploading
    if not os.path.exists(path):
        print(f"  Warning: File not found at '{path}', skipping.")
        continue

    # Upload the file
    # display_name helps identify the file in your Google Cloud project if needed
    response_file = client.files.upload(file=path, display_name=os.path.basename(path))
    uploaded_files.append(response_file)
    print(f"  Successfully uploaded: {path} as {response_file.name} ({response_file.display_name})")

if not uploaded_files:
    print("\nNo files were successfully uploaded. Exiting.")
else:
    # --- Content Generation ---
    print(f"\nGenerating content using model '{model_name}' with {len(uploaded_files)} image(s)...")

    # Construct the prompt content: list of uploaded file objects + text prompt
    # It's often good practice to put the text prompt last
    prompt_parts = uploaded_files + [text_prompt]
    # Call the generate_content method
    response = client.models.generate_content(
        model=model_name,
        contents=prompt_parts,
        # Optional: Add generation config for temperature, max output tokens etc.
        # generation_config=genai.types.GenerationConfig(
        #     temperature=0.4,
        #     max_output_tokens=2048
        # )
    )

    # Print the response text
    print("\n--- Response ---")
    print(response.text)
    print("---------------\n")


    # --- File Cleanup (Recommended) ---
    print("Cleaning up uploaded files...")
    for file_obj in uploaded_files:
        print(f"  Deleting: {file_obj.name} ({file_obj.display_name})")
        client.files.delete(name=file_obj.name)
        print(f"  Successfully deleted {file_obj.name}")

print("Script finished.")
```
